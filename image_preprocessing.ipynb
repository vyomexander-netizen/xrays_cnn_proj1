```python
import os
import cv2


outputdir = "dataset_processed"


folders = ["COVID19", "PNEUMONIA", "TUBERCULOSIS", "NORMAL"]

clahe = cv2.createCLAHE(   clipLimit=2.0,   tileGridSize=(8, 8))


for folder in folders:

    print("Processing:", folder)

    
    final = os.path.join(outputdir, folder)
    os.makedirs(final, exist_ok=True)


   
    for file in os.listdir(folder):

        
        input_path = os.path.join(folder, file)

        
        output_path = os.path.join(final, file)

        # Read image in grayscale
        img = cv2.imread(input_path, cv2.IMREAD_GRAYSCALE)

        if img is not None:

            # CLAHE
            img = clahe.apply(img)

            # Resize 
            img = cv2.resize(img, (224, 224))

            # Save  image
            cv2.imwrite(output_path, img)

          
```

```python
#split the data for train test validate
import os
import random
import shutil

inputdir = "dataset_processed"
outputdir = "dataset_split"

folders = ["COVID19", "PNEUMONIA", "TUBERCULOSIS", "NORMAL"]

for folder in folders:

    images = os.listdir(os.path.join(inputdir, folder))
    random.shuffle(images)

    train_length = int(0.7 * len(images))
    val_length = int(0.15 * len(images))
    test_length = int(0.15 * len(images))

    splits = {"train": images[:train_length],"val": images[train_length:train_length + val_length],"test": images[train_length + val_length:]}

    for split, split_images in splits.items():

        folderr = os.path.join(outputdir, split, folder)
        os.makedirs(folderr, exist_ok=True)

        for img in split_images:
            src = os.path.join(inputdir, folder, img)
            dst = os.path.join(folderr, img)

            shutil.copy(src, dst)

    
```
