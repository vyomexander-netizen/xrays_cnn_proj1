```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Lambda, Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.applications import DenseNet121
from tensorflow.keras.models import Model
```

```python
inputs = Input(shape=(224, 224, 1), name="xray_grayscale_input")


x = tf.keras.layers.Concatenate(axis=-1, name="grayscale_to_rgb")([inputs, inputs, inputs])

base_model = DenseNet121( weights="imagenet", include_top=False, input_tensor=x)


base_model.trainable = False
```

```python
x = GlobalAveragePooling2D(name="global_average_pooling")(base_model.output)
x = Dropout(0.3, name="classifier_help_overfitting")(x)
x = Dense(256, activation="relu", name="classifier_hidden_256")(x)

outputs = Dense(4, activation="softmax", name="classifier_output_4_classes")(x)

model = Model( inputs=inputs, outputs=outputs, name="DenseNet121_Xray_Classifier")
```

```python
model.save("DenseNet121_Xray_Classifier.keras")
```
